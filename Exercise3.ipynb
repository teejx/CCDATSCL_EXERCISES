Which type appears the most? Which appears the least?

```python
# put your answer here
all_types = pd.concat([df['Type 1'], df['Type 2']])
all_types = all_types.dropna()

# Count occurrences of each type
type_counts = all_types.value_counts()

print("Type counts:")
print(type_counts)

print("\nMost common type:")
print(type_counts.idxmax(), type_counts.max())

print("\nLeast common type:")
print(type_counts.idxmin(), type_counts.min())
```

How many unique primary types (Type 1) exist?

```python
# put your answer here
df['Type 1'].nunique()
```

How many unique secondary types (Type 2) exist?

```python
# put your answer here
df['Type 2'].nunique()
```


Which primary types have the most dual-type combinations?

```python
# put your answer here
dual_type_counts = df.groupby('Type 1')['Type 2'].nunique()
display(dual_type_counts.sort_values(ascending=False))
print('Water appears to be the most dual-type with 14 counts')
```

Which type has the highest mean Attack?

```python
# put your answer here
type_1_attack = df.groupby('Type 1')['Attack'].mean()
type_2_attack = df.groupby('Type 2')['Attack'].mean()

highest_type_1 = type_1_attack.sort_values(ascending=False).head(1)
highest_type_2 = type_2_attack.sort_values(ascending=False).head(1)

print("Highest for Type 1:")
print(highest_type_1)

print("\nHighest for Type 2:")
print(highest_type_2)

```

Which type has the lowest mean Defense?

```python
# put your answer here
type_1_defense = df.groupby('Type 1')['Defense'].mean()
type_2_defense = df.groupby('Type 2')['Defense'].mean()

highest_type_1 = type_1_defense.sort_values(ascending=False).head(1)
highest_type_2 = type_2_defense.sort_values(ascending=False).head(1)

print("Highest for Type 1:")
print(highest_type_1)

print("\nHighest for Type 2:")
print(highest_type_2)
```


Which generation has the highest average Speed?

```python
# put your answer here
avg_speed_per_generation = df.groupby('Generation')['Speed'].mean()
print(avg_speed_per_generation)
highest_speed_generation = avg_speed_per_generation.idxmax()
max_avg_speed = avg_speed_per_generation.max()
print(f"The generation with the highest average speed is Generation {highest_speed_generation} with an average speed of {max_avg_speed:.2f}")
```
